```python
#importing libraries 
import numpy as np
import pandas as pd
```

```python
#reading cafes sales data 
df = pd.read_csv('/workspaces/data_cleaning_project/data/raw/Cafe Sales.csv')
```

```python
df.head(10)
```

```python
df.info() # here most easy to handle is transaction date
```

```python
#checking if there are any duplicates records or not
print(f'cafe sales data contains {sum(df.duplicated())} duplicate records')
```

```python
#inspecting missing data from each column
df.isna().sum() * 100 / df.shape[0] #it will only  capture NaN values, not null or corrupts values like ERROR or Unknown
```

```python
#inspecting uniques values in all columns other thand trasaction id as that has 0 null values
df_cols = set(df.columns) # changing the column in set as set exclude repetative values and only saves unique value
exclude = {'Transaction ID'} #creates a new set
df_cols = df_cols.difference(exclude) # excluding transaction id by comparing two sets and saving the data in df_cols
df_cols


for col in df_cols:
    print(f'{col}: \n{df[col].unique()}', end = '\n\n')
```

```python
#replacing ERROR and UNKNOWN values with NaN
df = df.replace(['ERROR', 'UNKNOWN'],np.nan) # np.nan as it is related to numpy
```

```python
df.head(10)
```

```python
#changing data as int , float and date from str
df['Quantity'] = df['Quantity'].astype('Int64')
df[['Price Per Unit', 'Total Spent']] = df[['Price Per Unit', 'Total Spent']].astype('float')
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
```

```python
df.info()
```

```python
#inspecting missing data from each column after data treatment
df.isna().sum() * 100 / df.shape[0]
```

```python
#dropping records where transaction date is missing
df = df.dropna(subset =['Transaction Date'])
```

```python
df.info()
```

```python
#transaction_date is 0
df.isna().sum() * 100 / df.shape[0]
```

```python
#replacing values in payment method and location with INA(info not vaailable)
#dropping these high volume of records will lead to significant data loss
#inputing INA avoid bias
df[['Payment Method','Location']] = df[['Payment Method','Location']].fillna('INA')
```

```python
# null data is replaced with ina not null datat for payment method and location
df.isna().sum() * 100 / df.shape[0]
```

```python
#inspecting mean and median values for quantity
quantity_mean , quantity_median = round(df['Quantity'].mean() , 0) , round(df['Quantity'].median() , 0)

print(f'mean : {quantity_mean} & median : {quantity_median}')
```

```python
# since both are similar replacing missing values with mean values...median is better to use here both same so does not matter
df['Quantity'] = df['Quantity'].fillna(quantity_mean)
```

```python
# null data is replaced with mean values for quantity
df.isna().sum() * 100 / df.shape[0]
```

```python
# replacing missing values in Item with mode value, or 0 index value
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
```

```python
#null data of items are replaced
df.isna().sum() * 100 / df.shape[0]
```

```python
# checking how many distinct prices exist for each Item category and how many times each price point is registered.
df.groupby('Item')['Price Per Unit'].value_counts()
```

```python
 #filling missing Price Per Unit values with the most frequent (mode) price for each Item category
 df['Price Per Unit'] = df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
#removed null data of price per unit by replacing the data with mode value
df.isna().sum() * 100 / df.shape[0]
```

```python
# filling missing records for Total Spent by calculating the product of Quantity times Price Per Unit
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
```

```python
#all data are cleaaned data
df.isna().sum() * 100 / df.shape[0]
```
